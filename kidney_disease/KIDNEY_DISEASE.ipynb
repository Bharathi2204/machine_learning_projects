```python
def mean_values(df,column):
    mean_value=df[column].mean()
    df[column].fillna(value=mean_value,inplace=True)

def mode_values(df,column):
    mode_value=df[column].mode()[0]
    df[column].fillna(value=mode_value,inplace=True)
```

```python
num_col=[col for col in df.columns if df[col].dtypes!='object']
for col_name in num_col:
    mean_values(df,col_name)
```

```python
cat_col=[col for col in df.columns if df[col].dtypes=='object']
for col_name in cat_col:
    mode_values(df,col_name)
```

```python
df['Diabetes_Mellitus']=df['Diabetes_Mellitus'].replace(to_replace={' yes':'yes','\tno':'no','\tyes':'yes'})
df['Coronary_Artery_Disease']=df['Coronary_Artery_Disease'].replace(to_replace={'\tno':'no'})
df['CKD_Classification']=df['CKD_Classification'].replace(to_replace={'ckd\t':'ckd'})
```

```python
print(f"Diabetes_Mellitus : {df['Diabetes_Mellitus'].unique()}")
print(f"Coronary_Artery_Disease : {df['Coronary_Artery_Disease'].unique()}")
print(f"CKD_Classification : {df['CKD_Classification'].unique()}")
```

```python
df['CKD_Classification'] = df['CKD_Classification'].map({'ckd': 1, 'notckd': 0})
df['Red_Blood_Cells'] = df['Red_Blood_Cells'].map({'normal': 1, 'abnormal': 0})
df['Pus_Cell'] = df['Pus_Cell'].map({'normal': 1, 'abnormal': 0})
df['Pus_Cell_Clumps'] = df['Pus_Cell_Clumps'].map({'present': 1, 'notpresent': 0})
df['Bacteria'] = df['Bacteria'].map({'present': 1, 'notpresent': 0})
df['Hypertension'] = df['Hypertension'].map({'yes': 1, 'no': 0})
df['Diabetes_Mellitus'] = df['Diabetes_Mellitus'].map({'yes': 1, 'no': 0})
df['Coronary_Artery_Disease'] = df['Coronary_Artery_Disease'].map({'yes': 1, 'no': 0})
df['Appetite'] = df['Appetite'].map({'good': 1, 'poor': 0})
df['Pedal_Edema'] = df['Pedal_Edema'].map({'yes': 1, 'no': 0})
df['Anemia'] = df['Anemia'].map({'yes': 1, 'no': 0})
```

```python
print(f"Diabetes_Mellitus : {df['Diabetes_Mellitus'].unique()}")
print(f"Coronary_Artery_Disease : {df['Coronary_Artery_Disease'].unique()}")
print(f"CKD_Classification : {df['CKD_Classification'].unique()}")
```

```python
# Save preprocessed dataset
df.to_csv("processed_train.csv", index=False)
print("✅ Preprocessed dataset saved as 'processed_train.csv'")
```

```python
x=df.drop('CKD_Classification',axis=1)
y=df["CKD_Classification"]

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=42)

print("x_train: ",x_train.shape)
print("x_test: ",x_test.shape)
```

```python
#model building
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier

models=[]
models.append(("DecisionTreeClassifier",DecisionTreeClassifier()))
models.append(("Randomforest_classifier",RandomForestClassifier()))
models.append(("KNN_Classifier",KNeighborsClassifier(n_neighbors=5)))

print(models)
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score,confusion_matrix
```

```python
for name,model in models:
    print(f"model name: {name}")
    print("--------------------------------")
    model.fit(x_train,y_train)
    y_pred=model.predict(x_test)
    comparision=pd.DataFrame({"Actual_values":y_test.values,"predicted_values":y_pred}).head()
    print(comparision.to_string(index=False))
    print("\naccuracy_score = ",accuracy_score(y_test,y_pred))
    print('\n')
    print("confusion matrix \n",confusion_matrix(y_test,y_pred))
    print('\n')
    print("precision score = ",precision_score(y_test,y_pred))
    print('\n')
    print("recall_score = ",recall_score(y_test,y_pred))
    print('\n')
    print("f1_score = ",f1_score(y_test,y_pred))
    print('\n')


```

```python
import joblib

joblib.dump(models[0][1], "decision_tree_model.pkl")
joblib.dump(models[1][1], "random_forest_model.pkl")
joblib.dump(models[2][1], "knn_model.pkl")

print("All trained models have been saved successfully!")
```
